```python
l = [5,4,5,4,6,7,8,4,3,5,3,4]
unique_list = []
for element in l:
    if element not in unique_list:
        unique_list.append(element)
print(unique_list)
```

```python
l = [5,4,5,4,6,7,8,4,3,5,3,4]
freq_ele = {}
for element in l:
    if element in freq_ele:
        freq_ele[element] += 1
    else:
        freq_ele[element] = 1

print("Freq ele =", freq_ele)
```

```python
s = "I love python programming"
freq_char = {}
for ch in s:
    if ch in freq_char:
        freq_char[ch] += 1
    else:
        freq_char[ch] = 1

print("Freq char =", freq_char)
```
